```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This the 2x2 array: ")
print(array_2x2)

array_3x4 = np.random.rand(3,4)

print("Random 3x4 array: ")
print(array_3x4)

array_ones = np.ones((3,4))
array_zeroes = np.zeros((3,4))

print("Ones array: ")
print(array_ones)

print("Zeros array: ")
print(array_zeroes)

array_6x3 = np.random.rand(6,3)

print("Random 6x3 array: ")
print(array_6x3)

print("Transposed 6x3 array: ")
print(array_6x3.T)

n, p, m = 3, 4, 5
array_np = np.random.rand(n,p)
array_pm = np.random.rand(p,m)

print("Array np and pm dot product: ")
print(np.dot(array_np, array_pm))

print("Test commit")
```
